**3.LandSizeCateogory**

Categorizes land size into

1.Tiny

2.Small

3.Medium

4.Large.



```python
#Creating column
def land_size_category(x):
    if x <= 300:
        return 'Tiny'
    elif x <= 600:
        return 'Small'
    elif x <= 1000:
        return 'Medium'
    else:
        return 'Large'

df['LandSizeCategory']=df['Landsize'].apply(land_size_category)
```

**4.DistanceCategory**

Classified distance from the city center into categories:

1.Very Close

2.Close

3.Moderate

4.Far

5.Very Far

```python
#Creating column
def distance_category(x):
    if x < 10:
        return 'Very Close'
    elif x < 20:
        return 'Close'
    elif x < 30:
        return 'Moderate'
    elif x < 40:
        return 'Far'
    else:
        return '>Very Far'

df['DistanceCategory']=df['Distance'].apply(distance_category)
```

```python
#Display first 15 rows
df['DistanceCategory'].head(15)
```

**5.PriceCategory**

Categorizes properties based on price ranges:

1.Very Low

2.Low

3.Medium

4.High

5.Very High

```python
#Creating column
def price_category(x):
    if x < 500000:
        return 'Very Low'
    elif x < 1000000:
        return 'Low'
    elif x < 2000000:
        return 'Medium'
    elif x < 5000000:
        return 'High'
    else:
        return 'Very High'

df['PriceCategory']=df['Price'].apply(price_category)
```

**2.Average Price by SalesYear**

**Objective and methods used**

Evaluate the average property price per year to understand pricing trends.

1.Grouped data by SalesYear and calculated the mean property price.

2.Visualized trends using a line chart and a bar chart for comparison.

```python
df.groupby('SalesYear')['Price'].mean().apply(lambda x: f"{x:,.0f}")
```

**Visualization**

A line chart is used to show the trend of average property prices over time.
A bar chart is used as a supportive visualization to compare values between years.

```python
#Creating lineplot Line plot for trend using seaborn
avg_price_year=df.groupby('SalesYear')['Price'].mean().reset_index()

plt.figure(figsize=(8,5))
sns.lineplot(data=avg_price_year, x='SalesYear',y='Price',marker='o')

plt.title('Average Property Price by Year')
plt.xlabel('Year')
plt.ylabel('Average Price')

plt.show()

#Creatng barplot for comparison using seaborn
b_plot=sns.barplot(data=avg_price_year,x='SalesYear',y='Price',hue='SalesYear',palette='viridis',width=0.4,legend=False)

for container in b_plot.containers:
    b_plot.bar_label(container, fmt='%.0f')

plt.title('Average Property Price by Year (Comparison View)')
plt.xlabel('Year')
plt.ylabel('Average Price')

plt.show()
```

**Observations and Insights**

1.The average property price is almost similar in both years.

2.There is a slight decrease in average price in 2017 compared to 2016.
